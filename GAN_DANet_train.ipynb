# GAN-DANet Train Notebook

This notebook preserves the original GAN-DANet train entrypoint, but evolves it into a two-member hydrology ensemble.

The original strategy trained two models (`model1` and `model2`). This upgraded notebook keeps that structure for ensemble uncertainty, while using the temporal anomaly/trend-aware HydroGAN-DANet pipeline underneath.


```python
%matplotlib inline

from hydro_notebooks import load_notebook_config, train_notebook_ensemble

CONFIG_PATH = "hydro_experiment_config.yml"
TRAIN_OUTPUT_DIR = "outputs/notebooks/train"
ENSEMBLE_SEEDS = (42, 26)
NOTEBOOK_EPOCHS = None  # set an int to override the config file

config = load_notebook_config(CONFIG_PATH, output_dir=TRAIN_OUTPUT_DIR)
config
```

```python
train_summary = train_notebook_ensemble(
    config_path=CONFIG_PATH,
    output_dir=TRAIN_OUTPUT_DIR,
    seeds=ENSEMBLE_SEEDS,
    epochs=NOTEBOOK_EPOCHS,
)
train_summary
```

```python
import pandas as pd

member_rows = []
for member in train_summary["members"]:
    member_rows.append(
        {
            "member_index": member["member_index"],
            "seed": member["seed"],
            "checkpoint": member["compat_checkpoint"],
            "best_val_loss": member["summary"]["best_val_loss"],
            "best_epoch": member["summary"]["best_epoch"],
        }
    )

pd.DataFrame(member_rows)
```
